Main transformation start from here 


```python
# convert time to minute
import re 
def time_to_minutes(time):
    match=re.match(r'PT(?:(\d+)H)?:(\d+)M?',str(time))
    
    if not match:
        return 0
    hours=int(match.group(1)or 0)
    minutes=int(match.group(2) or 0 )
    return hours *60 + minutes 
    
```

```python
data['CookTime_minutes']=data['CookTime'].apply(time_to_minutes)
data['PrepTime_minutes']=data['PrepTime'].apply(time_to_minutes)
data['TotalTime_minutes']=data['TotalTime'].apply(time_to_minutes)
```

we'll parse Keywords , RecipeIngredientsParts . and RecipeIngredientsQunatities into porper python lists

```python
data.columns.tolist()
```

```python
import ast
def parse_list(value):
    try:
        value = str(value).strip()

        if value.startswith("c(") and value.endswith(")"):
            value = "[" + value[2:-1] + "]"

        result = ast.literal_eval(value)

        if isinstance(result, (list, tuple)):
            return list(result)

        if isinstance(result, str):
            return [result]

        return []

    except (ValueError, SyntaxError, TypeError):
        return []
list_cols = [
    "Keywords",
    "RecipeIngredientParts",
    "RecipeIngredientQuantities"
]

for col in list_cols:
    data[col] = data[col].apply(parse_list)
```

ingredient normalization. This is important because "Chicken Breast" and "chicken breast " should become the same ingredient.

```python
def clean_ingredients(ingredients):
    return[
        str(item).strip().lower()
        for item in ingredients
        if str(item).strip()
    ]
data['RecipeIngredientParts']=data['RecipeIngredientParts'].apply(clean_ingredients)
```
